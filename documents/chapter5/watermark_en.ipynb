# Dive into Large Language Models: Model Watermarking

Introduction: This section introduces language model watermarking

> Embed a "watermark" in content generated by a language model that is imperceptible to humans but detectable by algorithms.

## Tutorial Goals

1. Watermark embedding: embed a watermark when a language model generates content
2. Watermark detection: detect the watermark strength of a given text
3. Watermark evaluation: evaluate the detection performance of a watermarking method
4. Evaluate the robustness of the watermark (optional)

## Hands-on Example

> Use the KGW algorithm to embed a watermark in content generated by a language model

```JSON
{"prompt": "Ghost of Emmett Till: Based on Real Life Events "}
{"prompt": "Antique Cambridge Glass Pink Decagon Console Bowl Engraved Gold Highlights\n"}
{"prompt": "2009 > Information And Communication Technology Index statistics - Countries "}
......
```

- Each line is a JSON object and must contain at least a key named "prompt"
- The following content uses the `data/dataset/mc4/mc4.en.jsonl` file as an example. This file contains 500 data items in total; if model processing takes too long, you can reduce the data yourself.

### 3.2 Watermark Embedding

- Choose the model and watermarking algorithm. Here we choose the `baichuan-inc/Baichuan-7B` model and the `KGW` watermarking algorithm

  - ```Shell
    MODEL_NAME=baichuan-inc/Baichuan-7B
    MODEL_ABBR=baichuan-7b
    WATERMARK_METHOD_FLAG="--watermark_method kgw"
    ```

- Generate content and embed the watermark

  - ```Shell
    python3 gen.py \
        --base_model $MODEL_NAME \
        --fp16 \
        --batch_size 32 \
        --input_file data/dataset/mc4/mc4.en.jsonl \
        --output_file gen/$MODEL_ABBR/kgw/mc4.en.mod.jsonl \
        $WATERMARK_METHOD_FLAG
    ```

  - This command saves the model-generated content to the output file: `gen/$MODEL_ABBR/kgw/mc4.en.mod.jsonl`

### 3.3 Watermark Detection

> Watermark detection means: given a piece of text, compute its watermark strength (z-score).

- Compute the watermark strength of **watermarked** text

  - ```python
    python3 detect.py \
        --base_model $MODEL_NAME \
        --detect_file gen/$MODEL_ABBR/kgw/mc4.en.mod.jsonl \
        --output_file gen/$MODEL_ABBR/kgw/mc4.en.mod.z_score.jsonl \
        $WATERMARK_METHOD_FLAG
    ```

- Compute the watermark strength of **unwatermarked** text

  - ```python
    python3 detect.py \
        --base_model $MODEL_NAME \
        --detect_file data/dataset/mc4/mc4.en.jsonl \
        --output_file gen/$MODEL_ABBR/kgw/mc4.en.hum.z_score.jsonl \
        $WATERMARK_METHOD_FLAG
    ```

## Evaluating Watermark Robustness (Optional)

> After applying paraphrase and translation attacks to watermarked text, re-evaluate its detection performance

### 4.1 Preparation

We use the gpt-3.5-turbo-1106 model to paraphrase and translate the watermarked text. You may also choose other tools.

- Set the OpenAI API key

  - ```Shell
    export OPENAI_API_KEY=xxxx
    ```

- Modify RPM (requests per min) and TPM (tokens per min) in `attack/const.py`

### 4.2 Performing Attacks (using translation as an example)

- Translate the watermarked text into Chinese

  - ```Shell
    python3 attack/translate.py \
        --input_file gen/$MODEL_ABBR/kgw/mc4.en.mod.jsonl \
        --output_file gen/$MODEL_ABBR/kgw/mc4.en-zh.mod.jsonl \
        --model gpt-3.5-turbo-1106 \
        --src_lang en \
        --tgt_lang zh
    ```

- Re-evaluate

  - See [3.4](#eval)

- Compare the change in watermark performance before and after the attack
